# **Diferencias en el gradiente, estabilidad numérica y divergencias entre enfoques en regresión logística**

La regresión logística es un pilar fundamental en la clasificación supervisada. Aunque su formulación binaria es bien conocida, su extensión a problemas multiclase puede realizarse mediante dos enfoques principales: **Uno contra Todos (OvA)** y **regresión multinomial (Softmax)**. Estos métodos difieren no solo en su arquitectura, sino también en la forma en que se calcula el gradiente durante el entrenamiento.

---

## **Gradiente en la regresión logística binaria**

En la regresión logística binaria, el gradiente de la log-verosimilitud respecto al vector de parámetros $\theta$ es:

$$
\nabla_{\theta} l = X^{T}(y - p)
$$

donde $p$ es el vector de probabilidades predichas para la clase positiva. Este gradiente refleja el error entre las etiquetas reales y las probabilidades estimadas.

---

## **Enfoque Uno contra Todos (OvA)**

El enfoque OvA entrena un clasificador binario independiente por clase, tratando a cada una contra todas las demás. Cada clasificador tiene su propio vector de parámetros $\theta_k$ y su propio gradiente:

$$
\nabla_{\theta_k} l_k = X^{T}(y_k - p_k)
$$

donde $y_k$ es la codificación binaria para la clase $k$ y $p_k$ su probabilidad predicha.  
Los gradientes se calculan de forma independiente, sin interacción directa entre los clasificadores.

---

## **Regresión multinomial (Softmax)**

En contraste, la regresión multinomial (Softmax) entrena un único modelo con un vector de parámetros por clase $\theta = [\theta_1, \ldots, \theta_K]$.  
Su gradiente es una matriz:

$$
\nabla_{\theta} l = X^{T}(Y - P)
$$

donde $Y$ es la matriz de etiquetas en codificación one-hot y $P$ la matriz de probabilidades predichas por la función Softmax.  
Aquí, el gradiente de cada clase depende de las probabilidades de todas las clases, ya que la Softmax normaliza las puntuaciones para que sumen 1.  
Esta interdependencia es la diferencia clave: OvA optimiza de forma aislada, mientras que Softmax lo hace de forma conjunta.

---

## **Estabilidad numérica**

Esta normalización en la Softmax, aunque esencial, introduce un riesgo de inestabilidad numérica.  
Cuando las puntuaciones (logits) $z_k = \theta_k^{T} x$ son muy grandes, el cálculo de $\exp(z_k)$ puede provocar un desbordamiento (*overflow*).  
Para mitigarlo, se aplica una transformación de estabilidad numérica:

$$
z_k' = z_k - \max(z)
$$

antes de la exponenciación.  
Esta operación no altera las probabilidades resultantes, pero evita el *overflow*, ya que el máximo valor en el exponente será cero (Goodfellow, Bengio, & Courville, 2016).

---

## **Divergencias en predicciones**

A pesar de que ambos enfoques suelen producir resultados similares, pueden divergir en sus predicciones en ciertos escenarios.  
La principal causa es que OvA no garantiza que las probabilidades de las clases sumen 1, ya que cada clasificador opera de forma independiente.  
Esto puede llevar a situaciones en las que dos o más clases tengan una probabilidad predicha mayor que 0.5, generando ambigüedad en la decisión final.  
En cambio, la Softmax siempre produce una distribución de probabilidad válida.  
Esta diferencia es más pronunciada en conjuntos de datos con clases no linealmente separables o con distribuciones de clases muy desbalanceadas, donde la falta de coherencia global en OvA puede llevar a decisiones subóptimas (Bishop, 2006).

---

## **Conclusión**

La elección entre OvA y Softmax implica un compromiso entre simplicidad y coherencia probabilística.  
Mientras que OvA es intuitivo y modular, la Softmax, con su gradiente conjunto y su manejo cuidadoso de la estabilidad numérica, ofrece un marco más robusto y teóricamente sólido para la clasificación multiclase.
